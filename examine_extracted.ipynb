```python
import pandas as pd
from pathlib import Path

import os
```

```python
df = pd.read_csv("extracted_data/dataset.csv")
cols = ['Review_education', 'Review_gender', 'Review_income']
df = df[~df[cols].isin(['others', 'default']).any(axis=1)]
df = df.dropna(subset=cols).reset_index(drop=True)

for c in cols:
    print(c, df[c].unique())
```

```python
bins = [17, 25, 40, 60, 75]
labels = ['18-25', '26-40', '41-60', '61-75']
df['Review_age'] = pd.cut(df['Review_age'], bins=bins, labels=labels)
print(df['Review_age'].value_counts(dropna=True).sort_index())
```

```python
df.to_csv("extracted_final.csv", index=False)
```
